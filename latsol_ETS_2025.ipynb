```python
#no .1
bulan = input("Bulan apa anda ingin menginapa? (Mei, Juni, Juli, Agustus, September, Oktober)")
lama = int(input("Berapa Hari Anda Ingin Menginap?"))
tipe = input("Tipe kamar apa yang anda inginkan? studio/apartement:")

tarif_studio = 0
tarif_apartement = 0

if bulan == "Mei" or bulan == "oktober":
  tarif_studia = 5000000
  tarif_apartement = 650000
elif bulan == "Juni" or bulan == "September":
  tarif_studio = 752000
  tarif_apartement = 687000
elif bulan == "Juli" or bulan == "Agustus":
  tarif_studio = 760000
  tarif_apartement = 770000
else:
  print("Bulan yang anda masukkan tidak valid.")

total_tarif = 0
if tipe == "Studio":
  total_tarif = tarif_studio * lama
  if lama > 3:
    diskon = total_tarif * 0.15
    total_tarif -= diskon

total_apartement = 0
if tipe == "apartement":
  total_tarif = tarif_apartement * lama
if lama > 3:
  total_tarif = total_tarif - (0.10 * total_tarif)
else:
  print("Tipe kamar yang anda masukkan tidak valid")

if total_tarif > 0:
  print("Total tarif untuk menginap selama", lama, "Hari di bulan", bulan, "dengan tipe kamar", tipe, "adalah Rp", total_tarif)
```
