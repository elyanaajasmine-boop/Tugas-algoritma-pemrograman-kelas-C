```python
#no. 1
usia_pendaftar = int(input("Masukkan Umur: "))
nama_departemen = input ("Masukkan nama departemen: ")

cek = (usia_pendaftar >= 19) and ("Statistika" in nama_departemen)
print (f"Hasil Pengecekan: {cek}")
```

```python
#no. 2
data_pelanggan = ("khalila", "Premium")
berat_kardus = [25.3, 40.0, 35.2]

nama_pelanggan, status_member = data_pelanggan

total_berat = sum(berat_kardus)
cek_vip = (status_member == "Premium") or (total_berat > 100)
print (f"Hasil Pengecekan Jalur Pengiriman VIP: {cek_vip}")
```

```python
#no. 3
IPK = float(input("Masukkan IPK: "))
riwayat_nilai = ['C', 'B', 'AB', 'A']
lolos_beasiswa = (IPK >= 3.50) and ('E' not in riwayat_nilai)

print (f'Ketentuan Syarat Beasiswa: {lolos_beasiswa}')
```
